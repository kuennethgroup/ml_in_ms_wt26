# Prompting an LLM

Use https://www.perplexity.ai/, https://chat.openai.com/, https://gemini.google.com/, or https://claude.ai/.

### 1. Build a development strategy for your capstone / research project of this course

Objective: "fill research topic"

- Ask the model to explain the topic: "Explain ..."
- Get the model to collect more information that might be useful to solve the task: "Help me understand ..."
- Ask the model to do some task: "Write a proposal about ...", "Write Python code to solve ..."


**Example:**

Objective: Develop a machine learning model for the prediction of the glass transition temperature of polymers

1. Explain the glass transition temperature of polymers
2. Help me understand the correlation between the chemical structure of polymers and the glass transition temperature.
3. Help me understand the current data situation and find possible datasets with glass transition temperature and molecular weight.
4. Help me understand the impact of fingerprinting and data fusion on machine learning models for the prediction of glass transition temperature.
5. Write a proposal about developing a machine learning model for the prediction of the glass transition temperature of polymers.
6. Write Python code to solve this task.
7. Write Python code to solve this task using PyTorch and develop separate code to deploy this model using Streamlit. Make sure you save the model after finishing the training.


Do this for your own project. Save the input queries and try them with Perplexity, Gemini, OpenAI, etc.
